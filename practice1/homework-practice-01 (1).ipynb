## 2. Vizualizations

When working with data, it is often inconvenient to draw any conclusions if you look at a table and numbers in particular, so it is important to be able to visualize data. In this section we will do just that.

Matplotlib, of course, has [documentation](https://matplotlib.org/users/index.html) with a large number of [examples](https://matplotlib.org/examples/), but first it is enough to know about several main types of charts:
- plot - a regular point-by-point graph that can be used to depict curves or individual points;
- hist — a histogram showing the distribution of a certain value;
- scatter — a graph showing the relationship between two quantities;
- bar - a bar graph showing the relationship between a quantitative value and a categorical one.

In this task you will try to build each of them. It is equally important to learn the basic principles of visualization:
- the axes must be labeled on the graphs;
- the visualization must have a title;
- if several graphs are shown, an explanatory legend is required;
- all lines on the graphs must be clearly visible (there are no similar colors or colors that blend into the background);
- if a value is displayed that has an obvious range of values (for example, percentages can be from 0 to 100), then it is advisable to scale the axis over the entire range of values (the exception is the case when you need to show a small difference that is not noticeable at such scales).
- the grid on the graph helps to estimate the values ​​at points by eye, this is usually useful, so it is better to draw it.
